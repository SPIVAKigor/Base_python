```python
import json

# Пути к файлам (если они в той же папке, оставляем как есть)
input_purchases = 'purchase_log.txt'
input_visits = 'visit_log.csv'
output_funnel = 'funnel.csv'

### ЗАДАНИЕ 1: Загружаем покупки в оперативную память
purchases = {}
with open(input_purchases, 'r', encoding='utf-8') as f:
    next(f)  # Пропускаем заголовок
    for line in f:
        line = line.strip()
        dict_line = json.loads(line)
        # Сохраняем в словарь: ID пользователя -> Категория
        purchases[dict_line['user_id']] = dict_line['category']

print(f"Словарь покупок и количество записей: {len(purchases)}")


### ЗАДАНИЕ 2: Построчная обработка визитов и запись воронки
with open(input_visits, 'r', encoding='utf-8') as f_visits:
    with open(output_funnel, 'w', encoding='utf-8') as f_out:
        # Пропускаем заголовок в визитах, чтобы не искать категорию для 'user_id'
        next(f_visits)
        
        # Обрабатываем файл визитов 
        for line in f_visits:
            row = line.strip().split(',')
            user_id = row[0]
            
            # Проверяем наличие покупки для текущего пользователя
            category = purchases.get(user_id)
            
            if category:
                # Если покупка была, добавляем категорию и записываем в файл
                row.append(category)
                f_out.write(','.join(row) + '\n')

print(f"Файл '{output_funnel}'сохранен.")

# Проверка результата
print("\nПроверка первых строк результата:")
with open(output_funnel, 'r', encoding='utf-8') as check:
    for i, line in enumerate(check):
        print(line.strip())
        if i >= 4: break
```
