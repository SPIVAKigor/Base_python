# Задание 1
Напишите функцию, которая возвращает название валюты (поле ‘Name’) с максимальным значением курса с помощью сервиса www.cbr-xml-daily.ru...ly_json.js

```python
import requests
import numpy as np

def find_max_currency_name():
    url = "https://www.cbr-xml-daily.ru/daily_json.js"
    data = requests.get(url).json()

    currencies = data["Valute"]

    names = np.array([currencies[c]["Name"] for c in currencies])
    values = np.array([currencies[c]["Value"] for c in currencies])

    max_index = np.argmax(values) # смотрим на индекс с максимальным значением валюты
    return names[max_index]
```

```python
print(find_max_currency_name())
```

# Задание 2
Добавьте в класс Rate параметр diff (со значениями True или False), который в случае значения True в методах курсов валют (eur, usd итд) будет возвращать не курс валюты, а изменение по сравнению в прошлым значением. Считайте, self.diff будет принимать значение True только при возврате значения курса. При отображении всей информации о валюте он не используется.

```python
import requests

class Rate:
    def __init__(self, diff=False):
        self.diff = diff
        self.url = "https://www.cbr-xml-daily.ru/daily_json.js"

    def get_rates(self):
        return requests.get(self.url).json()["Valute"]

    def _get_currency(self, code):
        rates = self.get_rates()
        currency = rates[code]

        if self.diff:
            return currency["Value"] - currency["Previous"]
        else:
            return currency["Value"]

    def usd(self):
        return self._get_currency("USD")

    def eur(self):
        return self._get_currency("EUR")

    def get_all_rates(self):
        rates = self.get_rates()
        return rates
```

```python
rate = Rate(diff=True)
rate.get_all_rates()
```
